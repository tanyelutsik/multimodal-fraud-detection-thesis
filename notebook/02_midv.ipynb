```python
from pathlib import Path
import sys

project_root = Path.cwd().parent
sys.path.append(str(project_root / "src"))

from midv_metadata import collect_midv_metadata, save_metadata

midv_root = project_root / "data" / "midv"
output_path = project_root / "data" / "processed" / "midv_metadata.csv"

print("MIDV path:", midv_root)


df_midv = collect_midv_metadata(midv_root)
print(df_midv.head())
print(df_midv.shape)
print(df_midv[["width", "height"]].head())

save_metadata(df_midv, output_path)
```

The MIDV metadata were collected successfully from the original dataset structure.  
In total, 4,000 original document images were identified and labeled as "bona_fide".

At this stage, MIDV is used as a source of authentic document images.  
These samples will later serve two purposes:
1. as bona fide examples in the image classification task,
2. as source images for generating a separate suspicious class through controlled visual modifications.

For initial verification, suspicious sample generation was first tested on a small subset of 10 MIDV images.  
This step was used to confirm that the annotation paths, document-region extraction, visual transformations, and metadata saving worked correctly before applying the procedure to the full dataset.

As an initial verification step, the suspicious-image generation procedure was tested on a subset of 10 MIDV samples.

Out of these 10 input images, 8 suspicious variants were generated successfully.  
The difference between the number of input and output samples is expected, since some images may not contain usable annotation information for extracting the document region.

This test confirms that the generation pipeline works correctly and can produce suspicious MIDV samples together with the corresponding metadata.
