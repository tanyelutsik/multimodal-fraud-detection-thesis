```python
from sklearn.model_selection import train_test_split

full_df = pd.read_csv(processed_dir / "image_dataset_final.csv").copy()

group_df = (
    full_df.groupby("group_key", as_index=False)
    .agg({"image_class": "first"})
)

print("Unique groups:", len(group_df))
print(group_df["image_class"].value_counts())
```

```python
group_df = full_df[["group_key"]].drop_duplicates().copy()
print("Unique groups:", len(group_df))
print(group_df.head())
```

```python
train_groups, temp_groups = train_test_split(
    group_df["group_key"],
    test_size=0.30,
    random_state=42
)

val_groups, test_groups = train_test_split(
    temp_groups,
    test_size=0.50,
    random_state=42
)
```

```python
train_df_new = full_df[full_df["group_key"].isin(train_groups)].copy()
val_df_new = full_df[full_df["group_key"].isin(val_groups)].copy()
test_df_new = full_df[full_df["group_key"].isin(test_groups)].copy()
```

```python
train_set = set(train_df_new["group_key"])
val_set = set(val_df_new["group_key"])
test_set = set(test_df_new["group_key"])

print("Train-Val overlap:", len(train_set & val_set))
print("Train-Test overlap:", len(train_set & test_set))
print("Val-Test overlap:", len(val_set & test_set))
```

```python
for name, split_df in [("TRAIN", train_df_new), ("VAL", val_df_new), ("TEST", test_df_new)]:
    print(f"\n{name}")
    print("Rows:", len(split_df))
    print("Groups:", split_df["group_key"].nunique())
    print(split_df["image_class"].value_counts())
    print(split_df["image_class"].value_counts(normalize=True))
```

```python
train_set = set(train_df_new["group_key"])
val_set = set(val_df_new["group_key"])
test_set = set(test_df_new["group_key"])

print("Train-Val overlap:", len(train_set & val_set))
print("Train-Test overlap:", len(train_set & test_set))
print("Val-Test overlap:", len(val_set & test_set))
```
