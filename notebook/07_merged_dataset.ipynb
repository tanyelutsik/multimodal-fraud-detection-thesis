```python
from pathlib import Path
import pandas as pd

project_root = Path.cwd().parent
processed_dir = project_root / "data" / "processed"

print("Project root:", project_root)
print("Processed dir:", processed_dir)

mm_train = pd.read_csv(processed_dir / "mm_train.csv")
mm_val = pd.read_csv(processed_dir / "mm_val.csv")
mm_test = pd.read_csv(processed_dir / "mm_test.csv")

print("MM TRAIN:", mm_train.shape)
print("MM VAL:  ", mm_val.shape)
print("MM TEST: ", mm_test.shape)
```

```python
for name, df in [("MM TRAIN", mm_train), ("MM VAL", mm_val), ("MM TEST", mm_test)]:
    print(f"\n{name}")
    print(df["final_label"].value_counts())
    print(df["final_label"].value_counts(normalize=True) * 100)
```

```python
for name, df in [("MM TRAIN", mm_train), ("MM VAL", mm_val), ("MM TEST", mm_test)]:
    matches = (df["tab_isFraud"] == df["final_label"]).all()
    print(f"{name}: tabular labels match final_label -> {matches}")
```

```python
label_map = {
    "bona_fide": 0,
    "genuine": 0,
    "normal": 0,
    "clean": 0,
    "fake": 1,
    "fraud": 1,
    "forged": 1
}

for name, df in [("MM TRAIN", mm_train), ("MM VAL", mm_val), ("MM TEST", mm_test)]:
    mapped = df["img_image_class"].str.strip().str.lower().map(label_map)
    matches = (mapped == df["final_label"]).all()
    print(f"{name}: image labels match final_label -> {matches}")
```

```python
for name, df in [("MM TRAIN", mm_train), ("MM VAL", mm_val), ("MM TEST", mm_test)]:
    print(f"\n{name}")
    print("tab_split unique:", df["tab_split"].unique())
    print("img_split unique:", df["img_split"].unique())
    print("split unique:", df["split"].unique())
```
