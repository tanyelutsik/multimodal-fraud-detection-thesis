## FantasyID preprocessing

In this step, the raw FantasyID folder structure was converted into a structured metadata table. Each row in the table corresponds to one image and contains information such as file path, original train/test split, original folder category, capture type or attack type, image dimensions, and final class label.

The purpose of this preprocessing step was to make the dataset easier to analyze, standardize the label schema, and prepare the data for later merging with other image datasets.

For consistency with the thesis label definitions, the original FantasyID categories were mapped as follows:
- `bonafide` → `bona_fide`
- `attack` → `fraudulent`

During preprocessing, a small number of files stored in the `examples/` folder were found not to belong to the main class structure. Since these files could not be assigned a valid final label, they were excluded from the final metadata table.

The resulting cleaned metadata file was saved for later exploratory analysis and model training.

```python
from pathlib import Path
import sys

project_root = Path.cwd().parent
sys.path.append(str(project_root / "src"))

from fantasyid_preprocessing import build_fantasyid_dataframe, save_metadata

fantasy_root = project_root / "data" / "fantasyid"
output_path = project_root / "data" / "processed" / "fantasyid_metadata.csv"

print("FantasyID path:", fantasy_root)
print("Exists:", fantasy_root.exists())

df_fantasy = build_fantasyid_dataframe(fantasy_root)

print(df_fantasy.head())
print(df_fantasy.shape)
print(df_fantasy["final_label"].value_counts(dropna=False))
print(df_fantasy[df_fantasy["final_label"].isna()].shape)

save_metadata(df_fantasy, output_path)
```
