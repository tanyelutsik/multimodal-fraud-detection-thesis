```python
from pathlib import Path
import pandas as pd

project_root = Path.cwd().parent
processed_dir = project_root / "data" / "processed"

files_to_check = [
    "paysim_train.csv",
    "paysim_val.csv",
    "paysim_test.csv",
    "image_train_group_test.csv",
    "image_val_group_test.csv",
    "image_test_group_test.csv",
    "mm_train_group_test.csv",
    "mm_val_group_test.csv",
    "mm_test_group_test.csv",
]

for fname in files_to_check:
    path = processed_dir / fname
    print(fname, "->", path.exists())
```

```python
def summarize_image_split(name, path):
    df = pd.read_csv(path)
    print(f"\n{name}")
    print("Shape:", df.shape)
    print("Groups:", df["group_key"].nunique())
    print(df["image_class"].value_counts())
    return df

def summarize_mm_split(name, path):
    df = pd.read_csv(path)
    print(f"\n{name}")
    print("Shape:", df.shape)
    print(df["final_label"].value_counts())
    return df

paysim_train = pd.read_csv(processed_dir / "paysim_train.csv")
paysim_val = pd.read_csv(processed_dir / "paysim_val.csv")
paysim_test = pd.read_csv(processed_dir / "paysim_test.csv")

print("PAYSIM")
for name, df in [("train", paysim_train), ("val", paysim_val), ("test", paysim_test)]:
    print(f"\n{name}")
    print(df.shape)
    print(df["isFraud"].value_counts())

image_train = summarize_image_split("IMAGE TRAIN", processed_dir / "image_train_group_test.csv")
image_val = summarize_image_split("IMAGE VAL", processed_dir / "image_val_group_test.csv")
image_test = summarize_image_split("IMAGE TEST", processed_dir / "image_test_group_test.csv")

mm_train = summarize_mm_split("MM TRAIN", processed_dir / "mm_train_group_test.csv")
mm_val = summarize_mm_split("MM VAL", processed_dir / "mm_val_group_test.csv")
mm_test = summarize_mm_split("MM TEST", processed_dir / "mm_test_group_test.csv")
```

```python
mm_train_groups = set(mm_train["img_group_key"])
mm_val_groups = set(mm_val["img_group_key"])
mm_test_groups = set(mm_test["img_group_key"])

print("MM OVERLAP")
print("Train-Val:", len(mm_train_groups & mm_val_groups))
print("Train-Test:", len(mm_train_groups & mm_test_groups))
print("Val-Test:", len(mm_val_groups & mm_test_groups))
```
